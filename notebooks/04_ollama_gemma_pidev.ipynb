# Remesher Demo 04 — Ollama/Gemma + Pi + Remesher CLI skill

This notebook starts Ollama, pulls Gemma, and prepares Ollama's Pi integration to use `/workspace/remesher/skills/remesher-cli/SKILL.md`.

Modern Ollama can install and launch Pi directly with `ollama launch pi --model gemma4`. The RunPod image includes Node.js 22+ and npm because Ollama's Pi integration uses npm and current Pi packages require modern Node.


````python
import os, pathlib, subprocess, requests
WORKSPACE = pathlib.Path('/workspace')
ROOT = pathlib.Path('/workspace/remesher')
SKILL = ROOT / 'skills/remesher-cli/SKILL.md'
print(SKILL)
print(SKILL.read_text()[:1200])
````

````python
os.environ.setdefault('OLLAMA_MODEL', 'gemma4')
subprocess.run(['bash', 'docker/demo-jupyter/scripts/ollama-gemma4.sh'], cwd=ROOT, check=True)
````
