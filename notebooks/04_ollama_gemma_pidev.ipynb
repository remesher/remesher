# Remesher Demo 04 — Ollama/Gemma + pi.dev + Remesher CLI skill

This notebook starts Ollama, pulls Gemma, and prepares pi.dev to use `/workspace/remesher/skills/remesher-cli/SKILL.md`.

````python
import os, pathlib, subprocess, requests
ROOT = pathlib.Path('/workspace/remesher')
SKILL = ROOT / 'skills/remesher-cli/SKILL.md'
print(SKILL)
print(SKILL.read_text()[:1200])
````

````python
os.environ.setdefault('OLLAMA_MODEL', 'gemma4')
subprocess.run(['bash', 'docker/demo-jupyter/scripts/ollama-gemma4.sh'], cwd=ROOT, check=True)
````

````python
print(requests.get('http://127.0.0.1:11434/api/tags', timeout=10).json())
````

````python
# If this fails, set PIDEV_INSTALL to the official pi.dev install command and rerun.
subprocess.run(['bash', 'docker/demo-jupyter/scripts/install-pidev.sh'], cwd=ROOT, check=False)
````

````python
%%bash
set -euo pipefail
cd /workspace/remesher
export OLLAMA_HOST=http://127.0.0.1:11434
export OLLAMA_MODEL=${OLLAMA_MODEL:-gemma4}
export REMESHER_CLI_SKILL=/workspace/remesher/skills/remesher-cli/SKILL.md
export REMESHER_WORKDIR=/workspace/remesher
if command -v pi.dev >/dev/null 2>&1; then pi.dev --help; elif command -v pidev >/dev/null 2>&1; then pidev --help; elif command -v pi >/dev/null 2>&1; then pi --help; else echo 'pi.dev CLI not installed yet.'; fi
````
